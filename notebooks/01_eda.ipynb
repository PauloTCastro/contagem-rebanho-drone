# Análise Exploratória — Contagem de Rebanho por Drone

Explora o dataset antes do treino: quantas imagens, distribuição de animais por imagem, tamanho típico das caixas (crucial p/ decidir o tamanho do bloco no SAHI).

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

LBL_DIR = Path('../data/dataset/labels/train')

contagens, larguras, alturas = [], [], []
for txt in LBL_DIR.glob('*.txt'):
    linhas = [l.split() for l in txt.read_text().splitlines() if l.strip()]
    contagens.append(len(linhas))
    for _, _, _, w, h in linhas:
        larguras.append(float(w)); alturas.append(float(h))

print(f'Imagens: {len(contagens)}')
print(f'Animais/imagem -> media: {np.mean(contagens):.1f} | max: {max(contagens) if contagens else 0}')
print(f'Tamanho medio da caixa (normalizado): {np.mean(larguras):.4f} x {np.mean(alturas):.4f}')
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].hist(contagens, bins=30); ax[0].set_title('Animais por imagem')
ax[1].scatter(larguras, alturas, s=4, alpha=0.3)
ax[1].set_title('Dimensoes das caixas (norm.)'); ax[1].set_xlabel('largura'); ax[1].set_ylabel('altura')
plt.tight_layout(); plt.show()
```

**Como usar isto:** se as caixas forem muito pequenas em relação à imagem, o SAHI é essencial — reduza o tamanho do bloco (`--slice`) para dar mais resolução a cada animal.
